```python
#imports for whole assignment
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import sklearn as sk
from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
```

Task 2.1- Fisher Iris Dataset

The Iris Data set contains a np list array of data representing 3 plant species and its measurements.

the data contains the following attributes

1.   Samples    = 150 (50 per species)
2.   Features   = 4
      * Sepal Length
      * Sepal Width
      * Petal Length
      * Petal Width

The species of the sample is determined using the Iris_labels.npy combining the two places either a "0", "1", "2" next to the samples from the Iris_data.npy data set indentifying them as below



*   Species legend
      * 0 = Iris setosa (50 samples)
      * 1 = Iris versicolor (50 samples)
      * 2 = Iris virginica (50 samples)










```python
data1 = np.load("Iris_data.npy")
data2 = np.load("Iris_labels.npy")
data = np.column_stack((data1, data2))

#data
print(data1)
print(data1.size)
print(data1.shape)

#labels
print(data2)
print(data2.size)
print(data2.shape)

#combined
print(data)
```

```python
#species segmentation
setosa = data[data[:, 4] == 0]
versicolor = data[data[:, 4] == 1]
virginica = data[data[:, 4] == 2]
```

```python
plt.scatter(
    setosa[:, 0],
    setosa[:, 1],
    color="tab:purple",
    label="Setosa",
    edgecolor="k",
)
plt.scatter(
    versicolor[:, 0],
    versicolor[:, 1],
    color="tab:cyan",
    label="Versicolor",
    edgecolor="k",
)
plt.scatter(
    virginica[:, 0],
    virginica[:, 1],
    color="tab:orange",
    label="Virginica",
    edgecolor="k",
)

plt.title("Iris Data plot, (Sepal Length & Sepal Width in cm)")
plt.ylabel("sepal width (cm)")
plt.xlabel("sepal length (cm)")


plt.legend()
plt.show()
```

Task 2.2 - k-Means

K-Means initialisation

* Initialize the KMeans instance (using
k=3 and a set random_state for consistency)
* Checking data is being processed correctly

```python

X = np.array(data)
kmeans = KMeans(n_clusters=3, random_state=0, n_init="auto").fit(X)
print("Real Iris Clusters:")
print(kmeans.labels_)

print("\nReal Cluster Centers:")
print(kmeans.cluster_centers_)
```


* #1 Fit the model using ALL 4 feature dimensions
* #2 Predict cluster memberships
* #3 Extract the calculated cluster centroids
* #4 Scatter plot the predictions (using Sepal Length vs Sepal Width to compare with Task 2.1)

```python

#1
kmeans.fit(data)

#2
predicted_clusters = kmeans.predict(data)

#3
centroids = kmeans.cluster_centers_

#4
scatter_km = plt.scatter(
    data[:, 0],
    data[:, 1],
    c=predicted_clusters,
    cmap="plasma",
    edgecolor="k",
)

plt.scatter(
    centroids[:, 0],
    centroids[:, 1],
    color="red",
    marker="X",
    s=200,
    label="Centroids",
)

plt.title("K-Means Clustering Results (k=3)")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.legend()

plt.show()
```
